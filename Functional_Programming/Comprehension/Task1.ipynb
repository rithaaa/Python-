LIST COMPREHENSION

```python
# Create a list of *squares of all even numbers from 1 to 20* using list comprehension.

m=[i*i for i in range(1,20) if i%2==0]
print(m)
```

```python
# Given a list of words, create a new list containing only words with *more than 4 characters*.

l=['ritha','naja','food','cafe','hi']
m=[i for i in l if len(i)>=4]
print(m)
```

```python
#Given a list of numbers, create a list where *positive numbers remain the same and negative numbers are converted to 0*.

l=[12,-21,32,-14,-60]
m=[i if i>=0 else 0 for i in l]
print(m)
```

SET COMPREHENSION

```python
# Create a set of *squares of numbers from 1 to 10* using set comprehension.
m=set(i*i for i in range(1,10))
print(m)
```

```python
# Given a list of numbers, create a set containing *only the even numbers*.

l=[1,2,3,4,5,6,10,20,67,89]
m=set(i for i in l if i%2==0)
print(m)
```

```python
# Given a list of words, create a set containing the *first letter of each word*. 

l=['ritha','age','job']
m=set(i[0] for i in l)
print(m)
```

DICTIONARY COMPREHENSION

```python
# Create a dictionary where *numbers from 1 to 5 are keys and their squares are values*.

d={i:i*i for i in range(1,6)}
print(d)
```

```python
# Given a list of words, create a dictionary with *each word as the key and its length as the value*.

l=['ritha','hi','beautiful']
d={i:len(i) for i in l}
print(d)
```

```python
# Given numbers from 1 to 10, create a dictionary containing *only even numbers as keys and their cubes as values*.

q={i:i**3 for i in range(1,10) if i%2==0}
print(q)
```
