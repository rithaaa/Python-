```python
# w
f=open('func.txt','w')
f.write('Function is a reusable block of code which consist of set of instructions to perfome some specific task.')
f.close()
```

```python
# r
f=open('func.txt','r')
print(f.read())
```

```python
# r+
f=open('func.txt','r+')
print(f.read())
f.write('\nThere are two types:')
f.close()
```

```python
# a
f=open('func.txt','a+')
f.write('\nThey are inbuilt function and user-defined function.')
f.seek(0)
print(f.read())
f.close()
```
