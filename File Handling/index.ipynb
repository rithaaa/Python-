Two methods of opening,writing and closing the file

```python
# 1
v=open('text1.txt','w')
v.write('Ritha Fathima')
v.close()
```

```python
# 2
with open('text1.txt','w') as v:
    v.write('My name is Ritha')
```

```python
# writelines is used to enter collection of data
v=open('text1.txt','w')
v.writelines(['hi\n','baby\n','cake\n'])
v.close()
```

```python
# w+
# seek(index_value) is used to read the value from the index value that is given.
# seek is only used when we read the file after writing it.

v=open('text1.txt','w+')
v.write('hello i am ritha')
v.seek(0)
v.seek(3)
print(v.read())
```

READ OPERATION

r

```python
# read
v=open('text1.txt','r')
print(v.read())
```

```python
# readline
v=open('text1.txt','r')
print(v.readline())
```

```python
# readlines
v=open('text1.txt','r')
print(v.readlines())
```

```python
# r+    here the previous data will not be replaced
v=open('text1.txt','r+')
print(v.read())
v.write('\nI am a good girl.')
v.close()
```

APPEND

```python
# a    here appending only
v=open('text1.txt','a')
v.write("\nI am Append")
v.close()
```

```python
# a+ here write plus read 
v=open('text1.txt','a+')
v.write('\nHey i am 2nd append')
v.seek(0)
print(v.read())
v.close()
```
