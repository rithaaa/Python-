OOPS(Object Oriented Programming System)

```python
# w
o=open('oops.txt','w')
o.write('OOPS')
o.close()
```

```python
# r
o=open('oops.txt','r')
print(o.read())
```

```python
# r+
o=open('oops.txt','r+')
print(o.read())
o.write(":Object Oriented Programming System")
o.close()
```

```python
# a+
o=open('oops.txt','a+')
o.write('\nOOPS mainly focus on class and object.')
o.seek(0)
print(o.read())
o.close()
```
