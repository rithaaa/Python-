```python
# Variable length args
def display(*t):
    print(t)
    # print(type(*t))
display(2,3,4,5,6,7,8)
```

```python
# Variable length kwargs
def display(**t):
    print(t)
    
    # print(type(*t))
    
display(a=2,b=7,c=5)
```

```python
# Variable length kwargs&args
def display(*t,**a):
    print(t)
    print(a)
    print(type(*t))
    print(type(**a))
display(2,3,4,b=6,c=7,d=8)
```

```python
def display(a,b,c,d):
    print(a,b,c,d)
display(*[1,2,3,4])
```
