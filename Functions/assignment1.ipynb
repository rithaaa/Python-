```python
# 1.WAP to print numbers divisible by both 2 and 5 in the range 10 to 15.
def divisible():
    for i in range(10,16):
        if i%2==0 and i%5==0:
            print(i)
divisible()
```

```python
#2.WAP to print lowercase alphabets from z to a in reverse order.
def lower():
    for i in range(ord('z'),ord('a')-1,-1):
        print(chr(i),end=" ")
lower()
```

```python
#3.WAP to find the sum of odd numbers in the range 10 to 25.
def oddSum(sum=0):
    i=10
    while i<26:
        if i%2!=0:
            sum+=i
        i+=1
    print(sum)
oddSum()
```

```python
# 4.WAP to find the product of numbers in the range 1 to 5
def pro(pro=1):
    i=1
    while i<6:
        pro*=i
        i+=1
    print(pro)
pro()
```

```python
# 5.WAP to find the sum of the first n natural numbers.
def natSum(n,sum=0):
    i=1
    while i<=n:
        sum+=i
        i+=1
    print(sum)
natSum(5)
```

```python
# 6.WAP to print the first 3 numbers in the range 1 to 10.
def numbers(count=0):
    i=1
    while i<=10 and count<3:
        print(i)
        count+=1
        i+=1
numbers()
```

```python
# 7..WAP to print the last 3 numbers in the range 20 to 30.
def numbers(count=0):
    i=30
    while i>20 and count<3:
        print(i)
        count+=1
        i-=1
numbers()
```

```python
# 9.WAP to print all multiples of 5 between 1 and n.
def multiple(n):
    for i in range(1,n):
        if i%5==0:
            print(i,end=" ")
multiple(50)
```

```python
# 10.WAP to find the product of even individual digits in a given integer.
def pro(n,pro=1):
    while n>0:
        k=n%10
        if k%2==0:
            pro*=k
        n//=10
    print(pro)
pro(166)
```

```python
# 11.WAP to print every character of a given string.
def character(s):
    for i in s:
        print(i)
character('Morning')
```

```python
# 12.WAP to convert all lowercase characters in a string to uppercase.
def lowtoup(s,out=''):
    for i in s:
        if 'a'<=i<='z':
            out+=chr(ord(i)-32)
        else:
            out+=i
    print(out)
lowtoup('HeLLoMorning')
        
```

```python
# 13.WAP to convert uppercase characters to lowercase and lowercase characters to uppercase (swap case) in a given string.
def lowup(s,out=''):
    for i in s:
        if 'a'<=i<='z':
            out+=chr(ord(i)-32)
        elif 'A'<=i<='Z':
            out+=chr(ord(i)+32)
        else:
            out+=i
    print(out)
lowup('HeLLoMorning')
```

```python
# 14.WAP to find the product of all float numbers present at odd indices in a given tuple.
def floatodd(t,pro=1):
    for i in range(len(t)):
        if type(t[i])==float and i%2!=0:
            pro*=t[i]
    print(pro)
floatodd((1,2.4,1.3,5.7,[4,5],1.3))
    
```

```python
#1.WAP to find the sum of the largest and smallest digit in a given number.
def sum(n):
    n = str(n)
    largest = int(n[0])
    smallest = int(n[0])
    for i in range(1, len(n)):
        if largest < int(n[i]):
            largest = int(n[i])
        if smallest > int(n[i]):
            smallest = int(n[i])
    print(smallest + largest)

sum(143)
            
```

```python
# 2.WAP to check whether a given number is a Spy Number.
def spy(n,sum=0,product=1):
    while n>0:
        k=n%10
        sum+=k
        product*=k
        n//=10
    if sum==product:
        print("Spy Number")
spy(123)
```

```python
# 3.WAP to check whether a given number is a Buzz Number.
def buzz(n):
    if n%7==0 or n%10==7:
        print('Buzz Number')
    else:
        print("Not buzz number")
buzz(27)
buzz(11)
```

```python
# 4.WAP to find the factorial of each digit in a given number.
def factorial(n):
    fact=1
    for i in range(2,n+1):
        fact*=i
    print(fact)
factorial(5)
```
