CLASS OBJECT CONSTRUCTOR

```python
# Create a `Student` class with `name`, `age`, and `course` as attributes. Initialize these attributes using a constructor and create three student objects to display their details.

class Student:
    def __init__(self,name,age,course):
        self.name=name
        self.age=age
        self.course=course
    def show(self):
        print(self.name,self.age,self.course)
st1=Student('ritha',21,"python")
st1.show()
st2=Student('naja',21,"Java")
st2.show()
st3=Student('Hiba',27,"Web Tech")
st3.show()
```

INHERITANCE

```python
#  1. Create a `Vehicle` class containing `brand` and `price`. Create a `Car` class that inherits these properties and adds a `model` attribute. Display the complete details.

class Vehicle:
    brand='Toyota'
    price='40 lakh'
class Car(Vehicle):
    model='Innova'
c1=Car()
print(c1.brand)
print(c1.price)
print(c1.model)
```

```python
# 2. Create a `Person` class with `name` and `age` attributes. Create `Student` and `Teacher` classes that inherit these properties. Add `course` to `Student` and `subject` to `Teacher`.

class Person:
    name='ritha'
    age=21
class Student(Person):
    course='python'
class Teacher(Person):
    subject='Java'
s1=Student()
print(s1.name,s1.age,s1.course)
t1=Teacher()
print(t1.name,t1.age,t1.subject)

# class Person:
#     name='ritha'
#     age='21'
#     def display(self):
#         print(self.name,self.age)
# class Student(Person):
#     def course(self):
#         self.course='Python'
#         print(self.course)
#         print("")
# class Teacher(Person):
#     def subject(self):
#         self.subject='Java'
#         print(self.subject)
# s1=Student()
# s1.display()
# s1.course()
# t1=Teacher()
# t1.display()
# t1.subject()
```

```python
# Create a `Person` class with a `name` attribute. Create a `Student` class that adds `roll_no`, followed by a `CollegeStudent` class that adds `college_name`. Display all details.

class Person:
    name='Hiba'
class Student(Person):
    roll_no=47
class CollegeStudent(Student):
    college_name='MDIT'
p1=Person()
print(p1.name)
s1=Student()
print(s1.name,s1.roll_no)
c1=CollegeStudent()
print(c1.name,c1.roll_no,c1.college_name)
```

```python
# 4. Create a `Vehicle` class with a `brand` attribute. Create separate `Car` and `Bike` classes that inherit the property and add `doors` and `engine_cc` respectively.

class Vehicle:
    brand='Suzuki'
class Car(Vehicle):
    doors=4
class Bike(Vehicle):
    engine_cc=150
c1=Car()
print(c1.brand,c1.doors)
b1=Bike()
print(b1.brand,b1.engine_cc)
```
