```python
from abc import ABC,abstractmethod
class student_form(ABC):
    @abstractmethod
    def pan_no(self):
        pass
    def phone_no(self):
        pass
class st_details(student_form):
    def __init__(self,pan_no,phone_no):
        self.pan_car_number=pan_no
        self.phone_number=phone_no
    def pan_no(self):
        print(self.pan_car_number)
    def phone_no(self):
        print(self.phone_number)
student1=st_details(1234,987654321)
student1.pan_no()
student1.phone_no()
```

```python
from abc import ABC,abstractmethod
class scl_form(ABC):
    scl_n='ACS School'
    scl_l='Kochi'
    @abstractmethod
    def name(self):
        pass
    @abstractmethod
    def age(self):
        pass
    @abstractmethod
    def parent_name(self):
        pass
class st_details(scl_form):
    def __init__(self,name,age,parent_name):
        self.st_name=name
        self.st_age=age
        self.st_parent_name=parent_name
    def name(self):
        print(self.st_name)
    def age(self):
            print(self.st_age)
    def parent_name(self):
            print(self.st_parent_name)
    def display(self):
        print(self.scl_n)
        print(self.scl_l)
st1=st_details('ritha',21,'rajula')
st1.display()
st1.name()
st1.age()
st1.parent_name()
        
```

```python
# Create an abstract class Car with an abstract method start(), create BMW, Audi clases and implement start() differently.

from abc import ABC,abstractmethod
class car(ABC):
    @abstractmethod
    def start(self):
        pass
class BMW(car):
    def start(self):
        print("BMW Starting..")
class Audi(car):
    def start(self):
        print("Audi starting..")
b1=BMW()
b1.start()
a1=Audi()
a1.start()
```

```python
# CREATE AN ABSTRACT CLASS FOOD DELIVERY WITH AN ABSTRACT METHOD PLACE_ORDER(),CREATE SWIGGY AND ZOMATO CLASES AND IMPLEMENT THE METHOD DIFFERENTLY.

from abc import ABC,abstractmethod
class food_delivery(ABC):
    @abstractmethod
    def place_order(self):
        pass
class swiggy(food_delivery):
    def place_order(self):
        print("Swiggy")
class zomato(food_delivery):
    def place_order(self):
        print("zomato")
s1=swiggy()
s1.place_order()
z1=zomato()
z1.place_order()
    
```
