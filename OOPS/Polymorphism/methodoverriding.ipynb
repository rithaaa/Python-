```python
#Method Overriding

class animal:
    def sound(self):
        print("Every animal do sound")
class dog(animal):
    def sound(self):
        super().sound()
        print('Bow Bow')
husky=dog()
husky.sound()
```

```python
#Method overloading

#issue

class sums:
    def add(self,a,b):
        print(a+b)
    def add(self,a,b,c):
        print(a+b+c)
k=sums()
k.add(2,3,4)
# k.add(2,3)

#overcoming

class sums:
    def add(self,a=0,b=0,c=0):
        print(a+b+c)
s=sums()
s.add(2,3,4)
s.add(2,3)
```

```python


class sums:
    def add(self,*args):
        print((sum(args)))
s=sums()
s.add(2,3,4)
```
