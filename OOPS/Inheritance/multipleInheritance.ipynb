```python
# # MULTIPLE INHERITANCE

# 1. Employee Details
# Create PersonalDetails and JobDetails classes. Create an Employee class that inherits from both and displays the employee's complete information.

class PersonalDetails:
    def Per_detail(self,name,age):
        self.name=name
        self.age=age
class JobDetails:
    def job_d(self,job,sal):
        self.job=job
        self.sal=sal
class Employee(PersonalDetails,JobDetails):
    def display(self):
        print(self.name,self.age,self.job,self.sal)
p1=Employee()
p1.Per_detail('Ritha','21')
p1.job_d('Developer','70k')
p1.display()
    
```

```python
# 2. Student Information
# Create Student and Sports classes. Create a Result class that inherits from both and displays academic marks and sports score.

class Student:
    def academic(self,mark1,mark2):
        self.mark1=mark1
        self.mark2=mark2
class Sports:
    def sport(self,act1,act2):
        self.act1=act1
        self.act2=act2
class Result(Student,Sports):
    def display(self):
        print("Academic: ",self.mark1,self.mark2)
        print("Sports: ",self.act1,self.act2)
s1=Result()
s1.academic(21,45)
s1.sport('A','state-level')
s1.display()
```

```python
# 3. Smartphone
# Create Camera and Phone classes with suitable methods. Create a SmartPhone class that inherits from both and uses methods from both classes.

class Camera:
    def click(self,brand,megapixel):
        self.brand_c=brand
        self.mega=megapixel
class Phone:
    def Mobile(self,brand,color):
        self.brand=brand
        self.color=color
class SmartPhone(Camera,Phone):
    def display(self):
        print("Camera: ",self.brand_c,self.mega)
        print("Phone: ",self.brand,self.color)
s1=SmartPhone()
s1.click('Canon','450px')
s1.Mobile('Iphone','Black')
s1.display()
```

```python
# 4. Company Employee
# Create TechnicalSkills and CommunicationSkills classes. Create an Employee class that inherits from both and calculates/displays the employee's skills.

class TechnicalSkills:
    def tech(self,skill1,skill2):
        self.t_skill1=skill1
        self.t_skill2=skill2
class CommunicationSkills:
    def comm(self,skill1,skill2):
        self.c_skill1=skill1
        self.c_skill2=skill2
class Employee(TechnicalSkills,CommunicationSkills):
    def display(self):
        print("Tech: ",self.t_skill1,self.t_skill2)
        print("Comm: ",self.c_skill1,self.c_skill2)
e1=Employee()
e1.tech('Python','WebDevelopment')
e1.comm('Verbal','Writing')
e1.display()
```

```python
# 5. Online Order
# Create Product and Payment classes. Create an Order class that inherits from both and displays product details and payment information.

class Product:
    def pr(self,name,color):
        self.name=name
        self.color=color
class Payement:
    def pay(self,price,mode):
        self.price=price
        self.mode=mode
class Order(Product,Payement):
    def display(self):
            print("Product: ",self.name,self.color)
            print("Payment: ",self.price,self.mode)
o1=Order()
o1.pr('Shampoo','white')
o1.pay('500','Online')
o1.display()
```
