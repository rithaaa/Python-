```python
#  MULTILEVEL INHERITANCE

# 1. Employee Management
# Create Person → Employee → Manager.
# Store personal details in Person, employee details in Employee, and manager details in Manager. Display all information.

class Person:
    def p_det(self,name,age):
        self.name=name
        self.age=age
class Employee(Person):
    def e_det(self,job,sal):
        self.job=job
        self.sal=sal
class Manager(Employee):
    def m_det(self,m_id,dept):
        self.m_id=m_id
        self.dept=dept
    def display(self):
        print(self.name,self.age,self.job,self.sal,self.m_id,self.dept)
e1=Manager()
e1.p_det('Ritha','21')
e1.e_det('Developer','70k')
e1.m_det('101',5)
e1.display()
```
