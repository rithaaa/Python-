```python
# Single Inheritance

#1 Employee Salary
# Create a person class with name and age. Create an Employee class that inherit from person and adds employee ID and Salary. Display all details.

class Person:
    def p_n_a(self,name,age):
        self.name=name
        self.age=age
class Employee(Person):
    def e_id_sal(self,id,sal):
        self.id=id
        self.sal=sal
    def display(self):
        print(self.name,self.age,self.id,self.sal)
e1=Employee()
e1.p_n_a('ritha','21')
e1.e_id_sal('101','10k')
e1.display()  
    
```

```python
### 2. Bank Account

# Create an Account class with account number and balance. Create a SavingsAccount class that inherits from it and adds interest calculation.

class Account:
    def ac_bal(self,acc,bal):
        self.acct=acc
        self.bala=bal
class SavingsAccount(Account):
    def sav_acc(self,int_cal):
        self.int_cal=int_cal
    def show(self):
        print(self.acct,self.bala,self.int_cal)
s1=SavingsAccount()
s1.ac_bal('1234567890','45,000')
s1.sav_acc('60,000')
s1.show()
```

```python
### 3. Vehicle System

# Create a Vehicle class with brand and speed. Create a Car class that inherits from it and adds fuel type. Display complete car details.

class Vehicle:
    def br_sp(self,brand,speed):
        self.brand=brand
        self.speed=speed
class Car(Vehicle):
    def fuel(self,fuel_type):
        self.fuel_type=fuel_type
    def display(self):
        print(self.brand,self.speed,self.fuel_type)
v1=Car()
v1.br_sp('Toyota','120km/h')
v1.fuel('Petrol')
v1.display()
        
```

```python
#4 Student Result
# Create a Student class with name and roll number. Create a Result class that inherits from Student and calculates total and average marks.

class Student:
    def name_roll(self,name,roll):
        self.name=name
        self.roll=roll
class Result(Student):
    def cal(self,l):
        self.total=0
        for i in l:
            self.total+=i
        self.avg=self.total/len(l)
    def display(self):
        print(self.name,self.roll,self.total,self.avg)
s1=Result()
s1.name_roll('Ritha','47')
s1.cal([20,30,10])
s1.display()        
```

```python
#5 Product Management
# Create a Product class with product name and price. Create a DiscountedProduct class that inherits from it and calculates the final price after discount.

class Product:
    def name_price(self,name,price):
        self.name=name
        self.price=price
class DiscountedProduct(Product):
    def final(self,discount):
        self.final_price=self.price-discount
    def display(self):
        print(self.name,self.price,self.final_price)
p1=DiscountedProduct()
p1.name_price('Shirt',1000)
p1.final(200)
p1.display()
```
