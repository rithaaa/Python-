DECORATOR NORMAL


```python
# Write a program using a decorator that prints *"Welcome"* before calling a function.

def greetings(func):
    def welcome():
        print("Welcome")
        func()
    return welcome
@greetings
def e_name():
    print("Ritha Fathima")
e_name()

# With args

# def greetings(func):
#     def welcome(name):
#         print("Welcome")
#         func(name)
#     return welcome
# @greetings
# def name(name):
#     print(name)
# name("Ritha Fathima")
```

```python
# Write a program using a decorator that prints *"Start"* before and *"End"* after calling a function.

def signal(func):
    def s_e():
        print("START")
        func()
        print("END")
    return s_e
@signal
def programming():
    print("Basic proogramming")
    print("Advance Programming")
programming()    
```

DECORATOR USING *args AND **kwargs

```python
# Create a decorator that accepts a function with *any number of arguments* using *args and **kwargs.

def student(func):
    def details(*args,**kwargs):
        print("DETAILS:")
        func(*args,**kwargs)
    return details
@student
def st_details(name,age,status):
    print(name,age,status)
st_details('Ritha','21','Unemployed')
```

```python
# Write a decorator using *args that calls a function to **add 3 numbers** and prints a message before and after execution.

def operations(func):
    def add(*args):
        print("Addition")
        func(*args)
        print("Done",end="\n")
    return add
@operations
def add_op(a,b,c):
    print(a,b,c)
    print(a+b+c)
add_op(1,2,3) 
```
