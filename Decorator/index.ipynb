```python
def extra(func):
    def toppings():
        print('Choclate')
        func()
    return toppings

@extra
def geticecream():
    print("Ice Cream")
geticecream()
```

```python
def phone(func):
    def onoff():
        print("Phone ON")
        func()
        print("Phone OFF")
        print("")
    return onoff

@phone
def insta():
    print("Login")
    print("Insta opening")
    print("Logout")
    
@phone
def facebook():
    print("Login")
    print("Insta opening")
    print("Logout")

insta()
facebook()
```

```python
# Manager, Employee, CEO salary printing by chechking and checked

def check(func):
    def checking_sal():
        print("Checking")
        func()
        print('Checked')
        print(" ")
    return checking_sal
@check
def manager():
    print("Manager Salary : 2 Lakh")
    
@check
def ceo():
    print("CEO Salary: 10 Lakh")
@check
def employee():
    print("Employee Salary: 1 Lakh")

ceo()
manager()
employee()
```

```python
def operations(func):
    def result(*args,**kwargs):
        print("Do operations")
        func(*args,**kwargs)
        print("Done")
    return result
@operations
def add(a,b):
    print(a+b)
@operations
def sub(a,b):
    print(a-b)
add(2,3)
sub(2,3)
```
